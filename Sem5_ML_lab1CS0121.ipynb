Find sum of each line, mean, median and standard deviation of all lines in the file.

```
lines = []
with open("/poem.txt", "r") as poem:
    for line in poem:
        lines.append(line)
lengths = list(map(len, lines))
# print(lengths)
```

```
from functools import reduce
def mean(arr: list[int|float]):
    return sum(arr) / len(arr)
def median(arr: list[int|float]):
    arr.sort()
    n = len(arr)
    if n % 2 == 0:
        return int((arr[int(n/2)] + arr[int(n/2) - 1]) / 2)
    else:
        return arr[int(n/2)]
def variance(data: list[int|float]):
    mu = mean(data)
    sm = reduce(lambda x,y: x+y, list(map(lambda x: (x - mu)**2, data)))
    # sm = sum(*(map(lambda x: (x - mu)**2, data)))
    return sm / len(data)
def stdDeviation(data: list[float|int]):
    return variance(data) ** 0.5
```

```
print(f"mean: {mean(lengths)}")
print(f"median: {median(lengths)}")
print(f"standered Deviation: {stdDeviation(lengths)}")
```

Find the average cty fuel economy across all cars. All values in the dictionaries are strings.

```
# ans is the average of the numbers in the column with title cty
from functools import reduce
sm = reduce(lambda x,y: x+y, [float(row["cty"]) for row in mpg])
print(sm/len(mpg))
```

Return the unique values for the class types in our dataset using set.

```
vehicleclass = set([row["class"] for row in mpg])
vehicleclass
```

Find average miles per gallon in the highway for each class of vehicles

```
MileageByClass = []

for t in vehicleclass: # iterate over all the vehicle classes
    sumpg = 0
    classCount = 0
    for d in mpg:
        if d['class'] == t:
            sumpg += float(d["hwy"])
            classCount += 1
    MileageByClass.append((t, sumpg / classCount))

MileageByClass.sort(key=lambda x: x[1])
MileageByClass
```
